1.1 Liste Oluşturma ve Temel İşlemler

```python
vardiya_uretim = [1250, 1180, 1340]
print(f"vardiya retim listesi: {vardiya_uretim}")
```

```python
# append () methodu listenin sonuna eleman eklemeye yarar
vardiya_uretim.append(1290)
print(f'Ek vardiya eklendi:{vardiya_uretim}')
```

```python
# extend() methodu birden fazla veri eklemene yarar
hafta_sonu_uretim = [1100, 1050]
vardiya_uretim.extend(hafta_sonu_uretim)
print(f"Hafta Sonu Üretimi Eklendi: {vardiya_uretim}")
```

```python
# pop() methodu listeden son veriyi otomatik olarak kaldırır
son_vardiya = vardiya_uretim.pop()
print(f"Son vardiya çıkarıldı: {son_vardiya}")
print(f"Güncel Liste: {vardiya_uretim}")
```

# Alıştırma 2 -> borik asit ölçümü yapıldı ve bunun 99.62 olduğu ölçüldü. bunun H3BO3 değerine göre eğer ölçüm  urun spektinden büyükse uygundur değilse uygun değildir bunun raporunu yazdırınız.

```python
# Verilen veri seti
urun_spektleri = {
    'Etibor-48' : {
        'B2O3': {'min':48.0, 'max':49.35, 'birim':"%"},
        "Na2O": {'min':21.37, 'max':21.95, 'birim':"%"},
        "SO4": {'min':None, 'max':135, 'birim':"pm"},
        "Fe": {'min':None, 'max':3, 'birim':"ppm"}
    },
    'Borik Asit': {
        'H3BO3': {'min':99.9, 'max':None, 'birim':"%"},
        'B2O3': {'min':56.25, 'max':None, 'birim':"%"}
    },
    'Kolemanit':{
        'B2O3': {'min':42, 'max':50.9, 'birim':"%"},
        'CaO': {'min':24, 'max':27.2, 'birim':"%"} # Not: İkinci B2O3'ü CaO olarak düzelttim (Kolemanit yapısı gereği)
    }
}

# 1. Ölçüm değerini tanımlayalım
olcum_degeri = 99.62

# 2. Spekt değerini sözlükten çekelim
min_deger = urun_spektleri['Borik Asit']['H3BO3']['min']

# 3. Kontrol ve Raporlama
print(f"--- Borik Asit Analiz Raporu ---")
print(f"Ölçülen H3BO3: %{olcum_degeri}")
print(f"Beklenen Minimum: %{min_deger}")

if olcum_degeri >= min_deger:
    print("Sonuç: UYGUNDUR ✅")
else:
    print("Sonuç: UYGUN DEĞİLDİR ❌")
```

Bunu liste olarak yazın sonrasında 1200 hedefini tutturanların oranını raporlayın.
sadece hedefi tutturanların len() methodu ile listenin eleman sayısını ve sum() toplam üretilen tonu alacaksın

```python
liste = [1250, 1180, 1340, 1290, 1100, 1050, 1420]
boş_liste = []
hedef = 1200
for üretim_sayısı in liste:
    if üretim_sayısı>hedef:
        boş_liste.append(üretim_sayısı)
        eleman_sayısı = len(boş_liste)
        toplam_uretim= sum(boş_liste)
        oran= toplam_uretim/eleman_sayısı
print(f"Üretim hedefini geçenlerin raporu: {oran}")        
```
