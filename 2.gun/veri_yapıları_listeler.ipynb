## BÖLÜM 5: SPEKT KONTROL SİSTEMİ
### 5.1 Etibor-48 Kalite Kontrolü

```python
urun_adi = "Etibor-48"
olcum_b2o3 = 48.75
olcum_na2o = 21.50
olcum_so4 = 120
olcum_fe = 2

print(f"Ürün: {urun_adi}")
print(f"B₂O₃: {olcum_b2o3}%")
print(f"Na₂O: {olcum_na2o}%")
print(f"SO₄: {olcum_so4} ppm")
print(f"Fe: {olcum_fe} ppm")
```

```python
urun_spektleri
```

```python
spekt_b2o3 = urun_spektleri['Etibor-48']['B2O3']
b2o3_uygun = (olcum_b2o3 >= spekt_b2o3['min']) and (olcum_b2o3 <= spekt_b2o3['max'])

print(f"\nB₂O₃ Kontrolü:")
print(f"  Spekt: {spekt_b2o3['min']}-{spekt_b2o3['max']}{spekt_b2o3['birim']}")
print(f"  Ölçüm: {olcum_b2o3}{spekt_b2o3['birim']}")
print(f"  Durum: {'✓ UYGUN' if b2o3_uygun else '✗ UYGUN DEĞİL'}")
```

```python
# Diğer Parametreler
spekt_na2o = urun_spektleri[urun_adi]["Na2O"]
na2o_uygun = (olcum_na2o >= spekt_na2o["min"]) and (olcum_na2o <= spekt_na2o["max"])

spekt_so4 = urun_spektleri[urun_adi]["SO4"]
so4_uygun = olcum_so4 <= spekt_so4["max"]

spekt_fe = urun_spektleri[urun_adi]["Fe"]
fe_uygun = olcum_fe <= spekt_fe["max"]

print(f"Na₂O: {'✓' if na2o_uygun else '✗'}")
print(f"SO₄: {'✓' if so4_uygun else '✗'}")
print(f"Fe: {'✓' if fe_uygun else '✗'}")

tum_uygun = b2o3_uygun and na2o_uygun and so4_uygun and fe_uygun
print(f"\n{'='*40}")
print(f"SONUÇ: {'✓ TÜM TESTLER UYGUN' if tum_uygun else '✗ UYGUNSUZLUK'}")
print(f"{'='*40}")
```

# Bor Üretim Listesi Yönetimi

Bir bor madeni işletmesinde haftalık üretim miktarları (ton cinsinden) oluşturun. 

madde 1-) 7 elemanlı bir liste olucak üretim miktarları sırasıyla 1200, 1500, 1100, 1400, 1600, 1900,2000 ton olucak

madde 2-) Hafta içi üretimleri ayrı değişkene hafta_ici_uretim hafta sonu üretimlerini ayrı bir değişkene alalım hata_sonu_uretim. Bunun için dilimleme yapmalısınız

madde 3-) hafta içi üretimlerinin ortalaması almalısınız ve hafta sonu üretimlerinden ortamalasını almalısınız. sum() methoduyla toplama len() methoduyla eleman sayısını aldıktan sonra birbirine oranlayıp ortalamayı bulmalısınız.

madde 4-) ortalama 1600 tonun altına kalan değerler üretim yetersiz üstünde kalan değer için ise yeterli üretim diye rapor yazdırın.
tek satır if else çalıştırın.
örnek kod: print(f"SONUÇ: {'✓ TÜM TESTLER UYGUN' if tum_uygun else '✗ UYGUNSUZLUK'}")


```python
vardiya_uretim = [1200, 1500, 1100, 1400, 1600, 1900,2000]

hafta_ici_uretim = vardiya_uretim[:]
hafta_sonu_uretim = vardiya_uretim[-2:]

hafta_ici_ortalama_uretim = sum(hafta_ici_uretim) / len(hafta_ici_uretim)
hafta_sonu_ortalama_uretim = sum(hafta_sonu_uretim) / len(hafta_sonu_uretim)

esik_degeri = 1600

print(f"Hafta İçi Üretim -> {'Yeterli Üretim' if hafta_ici_ortalama_uretim >= esik_degeri else 'Yetersiz Üretim'}")

print(f"Hafta Sonu Üretim -> {'Yeterli Üretim' if hafta_sonu_ortalama_uretim >= esik_degeri else 'Yetersiz Üretim'}")

```
