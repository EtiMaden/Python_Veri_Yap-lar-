## 4. Aylık Tüvenan Cevher ve Konsantre Hesabı (14:00-14:30)

**Madencilik Terminolojisi:**
- **Tüvenan**: Ham cevher (ocaktan çıkan)
- **Konsantre**: İşlenmiş, zenginleştirilmiş ürün
- **Verim**: Tüvenandan konsantreye geçiş verimliliği

```python
tuvenan_ton = 12500.0 # ton (ham cevher)
tuvenan_b2o3 = 26.8 # % (tüvenan tenörü)

konsantre_ton = 3200.0 # ton (kosantre üretimi)
konsantre_b2o3 = 42.3 # % kosantre tenörü

# Verim = (Konsantre ton × Konsantre B₂O₃) / (Tüvenan ton × Tüvenan B₂O₃) × 100
verim = (konsantre_ton * konsantre_b2o3) / (tuvenan_ton * tuvenan_b2o3) * 100

# Toplam B2O3 miktarları
tuvenan_b2o3_ton = tuvenan_ton * (tuvenan_b2o3 / 100)
konsantre_b2o3_ton = konsantre_ton * (konsantre_b2o3 / 100)


# ===== f-STRING İLE DETAYLI RAPOR =====
print("=" * 70)
print(f"{'OCAK AYI ÜRETİM RAPORU':^70}")
print("=" * 70)
print()
print(f"{'TÜVENAN (HAM CEVHER)':-^70}")
print(f"  Toplam Tonaj        : {tuvenan_ton:>10,.2f} ton")
print(f"  Tenör (B₂O₃)        : {tuvenan_b2o3:>10.2f} %")
print(f"  İçerdiği B₂O₃       : {tuvenan_b2o3_ton:>10,.2f} ton")
print()
print(f"{'KONSANTRE (ZENGİNLEŞTİRİLMİŞ)':-^70}")
print(f"  Toplam Tonaj        : {konsantre_ton:>10,.2f} ton")
print(f"  Tenör (B₂O₃)        : {konsantre_b2o3:>10.2f} %")
print(f"  İçerdiği B₂O₃       : {konsantre_b2o3_ton:>10,.2f} ton")
print()
print(f"{'VERİMLİLİK':-^70}")
print(f"  Flotasyon Verimi    : {verim:>10.2f} %")
print(f"  Kayıp B₂O₃          : {tuvenan_b2o3_ton - konsantre_b2o3_ton:>10,.2f} ton")
print("=" * 70)
```

# Nem Yüzdesi + Kuru Baz

Nem Yüzdesi HesaplamasıKullanılan formül:$$\text{Nem \%} = \frac{(\text{Yaş Ağırlık} - \text{Kuru Ağırlık})}{\text{Yaş Ağırlık}} \times 100$$Verilen Değerler:Yaş Ağırlık: 150 kgKuru Ağırlık: 142 kgİşlem:Farkı bul: $150 - 142 = 8 \text{ kg}$ (Kaybedilen su miktarı)Oranla: $8 / 150 \approx 0,0533$Yüzdeye çevir: $0,0533 \times 100 = \textbf{\%5,33}$

```python
yas_kg = 150.0
kuru_kg = 142.0

nem_yuzde = ((yas_kg - kuru_kg) / yas_kg) * 100

print(f"Yaş: {yas_kg}kg Kuru: {kuru_kg} ve Nem %{nem_yuzde:.2f}")

# if nem_yuzde < 8:           # Rapor istersen çalıştır
#   print("Kabul Edilebilir")
# else:
#   print("Kurutma Gereklidir")
```

# Kuru Baz Tonaj (Ton Birimi)

$$\text{Kuru Tonaj} = \text{Yaş Tonaj} \times \left(1 - \frac{\text{Nem \%}}{100}\right)$$

Yaş Tonaj: 5,000 ton

Kuru Tonaj Bulacaksınız ve yaş tonajla farkını alıp kaç tonaj kaybedildiğini yazdıracaksınız.

$$\text{Kayıp Ton} = \text{Yaş Tonaj} - \text{Kuru Tonaj}$$

```python
yas_tonaj = 5000.0
kuru_tonaj = ((1-nem_yuzde/100) * yas_tonaj)

kayip_ton = yas_tonaj - kuru_tonaj


print(f"Yaş: {yas_tonaj} ton → Kuru: {kuru_tonaj:.2f} ton")
print(f"Kayıp: {kayip_ton:.2f} ton (%{nem_yuzde})")
```

# Kuru Baz Tenör Dönüşümü


Amaç:
bir bileşenin (örneğin $B_2O_3$) nemli bir karışımdaki yüzdesini, karışım kurutulduğunda ulaştığı yeni konsantrasyona dönüştürmeyi amaçlar. Nem uzaklaştığında toplam ağırlık azaldığı için, geriye kalan kuru maddedeki bileşen oranı artar.

Yaş Baz Değeri = 27.5

1. Kuru Bazdaki ($B_2O_3$) Oranının Hesaplanması:Nemden arındırılmış yeni konsantrasyonu bulmak için mevcut oranı kuru madde yüzdesine böleriz:$$\text{Kuru Baz} = \frac{\text{Yaş Baz}}{1 - (\text{Nem \%} / 100)}$$

$$\text{Artış} = \text{Kuru Baz} - \text{Yaş Baz}$$

```python
yas_b2o3 = 27.5

kuru_b2o3 = yas_b2o3 / (1- (nem_yuzde / 100))

artis = kuru_b2o3 - yas_b2o3

print(f"""
                      Tenor Raporu
      ---------------------------------------------------
                Yaş b2o3 -> {yas_b2o3}
                Kuru b2o3 -> {kuru_b2o3:.2f}
                Artış Tenor -> {artis:.2f} kadar olmuştur
    ------------------------------------------------------
""")
```
