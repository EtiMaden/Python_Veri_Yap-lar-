**"Peki ölçüm spekt dışındaysa, DÜŞÜK mü YÜKSEK mi bileceğiz?"**

👉 elif yapısına geçiş! 3 farklı durum:
1. Spekt içi
2. Spekt altı (düşük)
3. Spekt üstü (yüksek)

```python
b2o3 = 49.50
b2o3_min = 48.0
b2o3_max = 49.35

if b2o3 < b2o3_min:
    print(f"SPEKT DIŞI(DÜŞÜK): {b2o3}% < {b2o3_min}%")
    print("-> Konsantrasyon artırılacak")
elif b2o3 > b2o3_max:
    print(f"SPEKT DIŞI(BÜYÜK): {b2o3}% < {b2o3_min}%")
    print("-> Seyreltme yapılacak.")
else:
    print(f"SPEKT İÇİ {b2o3_min}% <= {b2o3}% <= {b2o3_max}")
    print("-> Sevkiyata hazır")
```

### 💡 ÖĞRETMEN SORUSU 3
**"Spekt sınırlarına çok yakın değerleri nasıl tespit ederiz?"**

**Yeni Kural:** 
- Sınıra 0.3% yakınsa → "SINIR DEĞER - Detaylı analiz gerekli" ⚠️

👉 Karar ağacı karmaşıklaştı! 5 durum var artık.

Görev:  
Python’da bir program yazarak:

Ölçüm değeri spekt minimumdan düşükse "❌ SPEKT DIŞI (DÜŞÜK)" yazdırınız.

Ölçüm değeri spekt maksimumdan yüksekse "❌ SPEKT DIŞI (YÜKSEK)" yazdırınız.

Ölçüm değeri min değerine çok yakınsa (tolerans içinde) "⚠️ SINIR DEĞER (MIN'E YAKIN)" uyarısı veriniz.

Ölçüm değeri max değerine çok yakınsa (tolerans içinde) "⚠️ SINIR DEĞER (MAX'A YAKIN)" uyarısı veriniz.

Ölçüm değeri güvenli aralıkta ise "✅ SPEKT İÇİ (GÜVENLİ BÖLGE)" yazdırınız.

```python
b2o3 = 49.30
ikaz_limiti = 0.3
b2o3_min = 48.0
b2o3_max = 49.35


if b2o3 < b2o3_min:
    print(f"SPEKT DIŞI(DÜŞÜK): {b2o3}% < {b2o3_min}%")
    print("-> Konsantrasyon artırılacak")
elif b2o3 > b2o3_max:
    print(f"SPEKT DIŞI(BÜYÜK): {b2o3}% < {b2o3_min}%")
    print("-> Seyreltme yapılacak.")
elif b2o3 < (b2o3_min + ikaz_limiti):
    print(f"⚠️ SINIR DEĞER (MIN'E YAKIN): {b2o3}%")
    print(f"   Min sınıra uzaklık: {b2o3 - b2o3_min:.2f}%")
    print("   → Detaylı analiz gerekli!")
elif b2o3 > (b2o3_max - ikaz_limiti):
    print(f"⚠️ SINIR DEĞER (MAX'A YAKIN): {b2o3}%")
    print(f"   Max sınıra uzaklık: {b2o3_max - b2o3:.2f}%")
    print("   → Detaylı analiz gerekli!")
else:
    print(f"SPEKT İÇİ (Güvenli Bölge)")
```

```python
# Gerçek laboratuvar ölçümleri
b2o3_olcum = 48.75
na2o_olcum = 22.0

# Etibor-48 spektleri
b2o3_min, b2o3_max = 48.0, 49.35
na2o_min, na2o_max = 21.37, 21.95


b2o3_uygun = b2o3_min <= b2o3_olcum <= b2o3_max
if b2o3_uygun:
    print(f"B2O3: {b2o3_olcum}% - UYGUN")
else:
    print(f"B2O3 {b2o3_olcum}%- Uygun Değil")


# n2o3 için kontrol
na2o_uygun = na2o_min <= na2o_olcum <= na2o_max
if b2o3_uygun:
    print(f"B2O3: {na2o_olcum}% - UYGUN")
else:
    print(f"B2O3 {na2o_olcum}%- Uygun Değil")


# Genel Karar Etibol-34 ürünü için
if b2o3_uygun and na2o_uygun:
    print('Ürün Sevkiyata hazır')
else:
    print('Ürün Yeniden İşlenecek')
    if not b2o3_uygun:
        print(f" -> b2o3 düzeltilmeli")
    if not na2o_uygun:
        print(f" -> na2o düzeltilmeli")
```

### 🧪 BÜYÜK ALIŞTIRMA 1.4: GERÇEK SPEKT KONTROL SİSTEMİ

**Borik Asit için 3 parametre kontrol edin:**
- H₃BO₃: min 99.9%
- SO₄: max 300 ppm
- Cl: max 7 ppm

**Ölçümler:**
- H₃BO₃: 99.92%
- SO₄: 280 ppm
- Cl: 5 ppm

Her parametreyi kontrol edin, sonra genel karar verin!

Python’da bir program yazarak ölçüm değerlerini spekt değerleriyle karşılaştırınız.

Eğer H₃BO₃ min değerinden düşükse uyarı veriniz.

Eğer SO₄ veya Cl max değerinden yüksekse uyarı veriniz.

Aksi durumda “✅ UYGUN” çıktısı veriniz.

```python
# ÖĞRENCİLER YAPSIN - Borik Asit Kontrol
# Spektler
h3bo3_min = 99.9
so4_max = 300
cl_max = 7

# Ölçümler
h3bo3_olcum = 99.0
so4_olcum = 310
cl_olcum = 9

print("🔬 BORİK ASİT KALİTE KONTROLÜ")
print("="*50)


if h3bo3_olcum < h3bo3_min:
    print(f"H3BO Yetersiz: {h3bo3_olcum}")
else:
    print(f"H3BO3 Yeterli: {h3bo3_olcum}")

if so4_olcum > so4_max:
    print(f"SO4 Limit Aşıldı: {so4_olcum}")
else:
    print(f"SO4 uygun: {so4_olcum}")

if cl_olcum > cl_max:
    print(f"CL Limit Aşıldı: {cl_olcum}")
else:
    print(f"CL uygun: {cl_olcum}")


h3bo3_uygun = h3bo3_olcum > h3bo3_min
so4_uygun = so4_olcum < so4_max
cl_uygun = cl_olcum < cl_max


# GENEL KARAR VEREN YERİ YAZIN 
# ÖRNEK KOD
# Genel Karar Etibol-34 ürünü için

print()
if h3bo3_uygun and so4_uygun and cl_uygun:
    print('Ürün Sevkiyata hazır')
else:
    print('Ürün Yeniden İşlenecek')
    if not h3bo3_uygun:
        print(f" -> h3bo3 düzeltilmeli")
    if not so4_uygun:
        print(f" -> so4 düzeltilmeli")
    if not cl_uygun:
        print(f"-> cl düzeltilmeli")
```
