## 3.4 `while` Döngüsü — Koşula Bağlı Tekrar

Bir AOI'de hızın belirli bir eşiği aşıp aşmadığını **simülasyon** üzerinden kontrol edelim. Hız, saatte 0.5 mm/saat artıyor olsun ve KIRMIZI eşik 20 mm/saat olsun.

````python
hiz = 1.0 # başlangıç hızı (mm/saat)
kirmizi_esik = 20.0  # kırmızı alarm eşiği
saat = 0

print("İvmenlenme similasyonu - Kırmızı Eşiğe ulaşana kadar")

while hiz < kirmizi_esik:
    print(f" Saat {saat:2d} -> {hiz:2f} mm/saat toprak kayma hızı vardır eşiğin altındadır.")
    hiz += 0.5 # hiz = hiz + 0.5
    saat += 1 # saat = saat + 1

print(f"\n Saat{saat} -> hız = {hiz:2f} mm/saat -> KIRMIZI ALARM tetiklendi")
````

````python
motor_verileri = [
    (45, 0.95), (48, 0.88), (52, 0.20), (50, 0.90),
    (92, 0.98),  # ← KRİTİK ISI: Motor çok sıcak!
    (47, 0.85), (46, 0.80), (44, 0.15)
]

guvenilir_olcumler = []

guven_esigi= 0.50  # 0.50 altı güvenilmez
kritik_sicaklik= 85     # 85 derece üstü tehlikeli

for olcum_no, (sicaklik, guven) in enumerate(motor_verileri,1):
    if guven < guven_esigi:
        print(f"Ölçüm No {olcum_no:2f} -> Güvenilirlik {guven:2f} ÇOK DÜŞÜK, atlandı")
        continue # bu pikseli analiz etme sıradakine geç
    guvenilir_olcumler.append(sicaklik)

    if sicaklik > kritik_sicaklik:
        print(f"Ölçüm no {olcum_no:2f} -> {sicaklik:2f} mm Aşırı Sıcaklık Uyarısı.")
        break # büyük hareket bulunud başka piksele bakmaya gerek yok

print(f"\n Güvenilir sıcaklık ölçüm sayısı : {len(guvenilir_olcumler)}")
print(f" Güvenilir Ölçümler(mm) : {guvenilir_olcumler}")
````

Egzersiz: SSR Alarm Seviyesi Fonksiyonu
Aşağıda bir SSR (Yüzey Şekil Değişimi Radarı) ölçüm sistemine ait veriler verilmiştir:

Görev: alarm_seviyesi(hiz_mm_saat) adlı bir fonksiyon yazınız. Bu fonksiyon aşağıdaki kurallara göre 'NORMAL', 'SARI', 'TURUNCU' veya 'KIRMIZI' döndürmelidir:

Hız tabanlı kurallar:

Hız (mm/saat)	Seviye
> 20.0	KIRMIZI
> 5.0	TURUNCU
> 1.0	SARI
≤ 1.0	NORMAL
Kümülatif yer değiştirme kuralları (kumulatif_mm verilmişse ek olarak uygulayın):


  AOI_A    → hız=  0.50 mm/saat→ NORMAL
  AOI_B    → hız=  2.30 mm/saat → SARI
  AOI_C    → hız=  7.80 mm/saat → TURUNCU
  AOI_D    → hız= 25.00 mm/saat → KIRMIZI
Fonksiyonunuzu yazdıktan sonra aoi_verileri sözlüğündeki her AOI için sonuçları yazdırınız.

Kümülatif yer değiştirme kuralları (kumulatif_mm verilmişse ek olarak uygulayın):

kumulatif_mm > 60 ise seviye direkt KIRMIZI olur.
kumulatif_mm > 30 ve mevcut seviye NORMAL olucak


fonskiyon ismi 
alarm_seviyesi(0.5, 62) -> KIRMIZI
alarm_seviyesi(6.0,35) -> NORMAL
alarm_seviyesi(22.0, 25) -> Kırmızı

````python
def alarm_seviyesi(hiz_mm_saat,yer_degistirme_mm=None):
    """SSR ölçümüne göre alarm seviyesini belirler.
    
    Parametreler:
        hiz_mm_saat (float): Yer değiştirme hızı, mm/saat.
        kumulatif_mm (float, isteğe bağlı): Kümülatif yer değiştirme, mm.
                                            Verilirse hız ile birleşik karar verilir.
    Döndürür:
        str: 'NORMAL', 'SARI', 'TURUNCU' veya 'KIRMIZI'
    """
    # Hız tabanlı temel karar
    if hiz_mm_saat > 20.0:
        seviye = "KIRMIZI"
    elif hiz_mm_saat > 5.0:
        seviye = "TURUNCU"
    elif hiz_mm_saat > 1.0:
        seviye = "SARI"
    else:
        seviye = "NORMAL"

    if yer_degistirme_mm is not None and yer_degistirme_mm > 60.0:
        seviye = 'KIRMIZI'
    elif yer_degistirme_mm is not None and yer_degistirme_mm > 30.0:
        seviye = 'NORMAL'
    else:
        seviye = seviye

    return seviye

aoi_verileri = {
    "AOI_A": {"hiz": 0.5,  "yer_degistirme": 10.0},
    "AOI_B": {"hiz": 2.3,  "yer_degistirme": 35.0},
    "AOI_C": {"hiz": 7.8,  "yer_degistirme": 25.0},
    "AOI_D": {"hiz": 25.0, "yer_degistirme": 70.0},
}

for bolge, parametreler in aoi_verileri.items():
    seviye = alarm_seviyesi(parametreler['hiz'], parametreler['yer_degistirme'])
    print(f" Bolge {bolge} -> hiz = {parametreler['hiz']} mm/saat km = {parametreler['yer_degistirme']}mm -> {seviye}")
````

````python
# size verilen [45.1, 47.8, 46.3, 48.9, 47.0, 48.1, 51.2, 46.5, 47.6, 49.0] bu listedeki değerleri aoi_istatistik isimli
# fonksiyon içerisinden sorted() methoduyla sıralama yapıp değişkene atayın
# eleman sayısını alaarak bir değişkene atayın len()
# ortalamasını alarak bir değişkene atayın
# en büyük değeri ayrı bir değişkene en küçük değeri ayrı bir değişkene atayın
# hepsini return edin tek satırda
````

````python
def aoi_istatistik(piksel_listesi):

    if not piksel_listesi:
        return (0.0, 0.0, 0.0, 0.0)
    
    sirali = sorted(piksel_listesi)
    n = len(sirali)
    ortalama = sum(sirali) / n
    en_kucuk = sirali[0]
    en_buyuk = sirali[-1]

    return (sirali,n,ortalama,en_kucuk,en_buyuk)


aoi_K2_pikselleri = [45.1, 47.8, 46.3, 48.9, 47.0, 48.1, 51.2, 46.5, 47.6, 49.0]


sirali_liste,adet,ortalama,en_kucuk,en_buyuk =  aoi_istatistik(aoi_K2_pikselleri)

print(f"""
                                AOI İSTATİSTİKLERİ

        Büyükten Küçüğe Sıralanmış Liste -> {sirali_liste}
        Adet Sayısı: -> {adet}
        Ortalama -> {ortalama}
        En Küçük Değer -> {en_kucuk}
        En Büyük Değer -> {en_buyuk}
""")
````

````python
def hiz_hesapla(yer_deg_baslangic, yer_deg_bitis, sure_saat):
    """İki ölçüm arasındaki ortalama hızı (mm/saat) hesaplar."""
    if sure_saat <= 0:
        return 0.0
    return (yer_deg_bitis - yer_deg_baslangic) / sure_saat


def ters_hiz(hiz_mm_saat):
    """Ters hız (1/V) — kılavuz 2.3 Saito modeli.
    
    Sıfıra yaklaşan ters hız → yakın göçme habercisi.
    """
    if hiz_mm_saat <= 0:
        return float('inf')   # hız 0 ise sonsuz (göçme yok)
    return 1.0 / hiz_mm_saat


def saito_analizi(yer_deg_baslangic, yer_deg_bitis, sure_saat):
    """Komposizyon: hem hızı hem ters hızı hesaplar ve yorumla döndürür."""
    h = hiz_hesapla(yer_deg_baslangic, yer_deg_bitis, sure_saat)
    th = ters_hiz(h)
    
    if h <= 0:
        yorum = "Hareket yok veya geri dönüş"
    elif th > 1.0:
        yorum = "Yavaş hareket — birincil/ikincil creep"
    elif th > 0.1:
        yorum = "İvmelenme başladı — yakın takip"
    else:
        yorum = "ÜÇÜNCİL creep — KIRMIZI alarm değerlendir"
    
    return h, th, yorum


# Üç farklı AOI üzerinde test
test_durumlari = [
    ("AOI_K1 (sakin)",       45.0, 45.8, 24),
    ("AOI_K2 (turuncu)",     20.0, 47.2, 12),
    ("AOI_GD1 (kritik)",     0.0,  68.5, 3),
]

for ad, basla, bit, sure in test_durumlari:
    h, th, yorum = saito_analizi(basla, bit, sure)
    print(f"{ad:25s}")
    print(f"   Hız     : {h:6.2f} mm/saat")
    print(f"   Ters Hız: {th:6.3f}  →  {yorum}")
    print()
````
