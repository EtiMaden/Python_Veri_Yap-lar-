## BÖLÜM 1: LİSTELER VE TUPLE YAPILARI
### 1.1 Liste Oluşturma ve Temel İşlemler

```python
vardiya_uretim = [1250, 1180, 1340]
print(f"Vardiya üretim listesi: {vardiya_uretim}")
```

```python
# append() methodu listenin sonuna eleman eklemeye yarar
vardiya_uretim.append(1290)
print(f'Ek vardiya eklendi:{vardiya_uretim}')
```

```python
#extend() methodu birden fazla veri eklemene yarar
hafta_sonu_uretim = [1100, 1050]
vardiya_uretim.extend(hafta_sonu_uretim)
print(f"Hafta Sonu Üretimi Eklendi: {vardiya_uretim}")
```

```python
#pop() methodu listeden son veriyi otomatik olarak kaldırır
son_vardiya = vardiya_uretim.pop()
print(f"Son vardiya çıkarıldı: {son_vardiya}")
print(f"Güncel Liste: {vardiya_uretim}")
```

# İndexleme ve Dilimleme

```python
ilk_uc_vardiya = vardiya_uretim[:3] # [:3] :'nın soluna bir şey yazılmadığı vakit varsıyalan olarak en baştaki değeri alır
print(f"İlk 3 vardiya: {ilk_uc_vardiya}")
```

```python
son_iki_vardiya = vardiya_uretim[4:]# [4:] :'nın sağ kısmı boş bırakıldığında varsayılan olarak en sonun indeks dahil alır
print(f"son_iki_vardiya: {son_iki_vardiya}")
```

```python
cift_vardiyalılar = vardiya_uretim[::2] # [::2] buradaki 2 değeri 2 şer 2 şer atlayarak gideceğini gösterir.
print(f"Çift vardiylaların üretimi: {cift_vardiyalılar}")
```

# İstatiksel Özet

```python
toplam = sum(vardiya_uretim)
ortalama = toplam / len(vardiya_uretim) # len() eleman sayısını verir
minimum = min(vardiya_uretim)
maximum = max(vardiya_uretim)

print(f"Toplam: {toplam} ton")
print(f"Ortalam: {ortalama} ton")
print(f"Minimum: {minimum} ton")
print(f"Maksimum: {maximum} ton")
```

# Sıralama

```python
vardiya_kopya = vardiya_uretim.copy() # kopyasını aldık orjinalliği bozmamak için

vardiya_kopya.sort() # küçükten büyüğe doğru sıralar
print(f"Artan Sıralama: {vardiya_kopya}")


vardiya_kopya.sort(reverse=True) # büyükten küçüğe doğru sıralar
print(f"Azalan Sıralama: {vardiya_kopya}")
```

## BÖLÜM 5: SPEKT KONTROL SİSTEMİ
### 5.1 Etibor-48 Kalite Kontrolü

```python
urun_adi = "Etibor-48"
olcum_b2o3 = 48.75
olcum_na2o = 21.50
olcum_so4 = 120
olcum_fe = 2

print(f"Ürün: {urun_adi}")
print(f"B₂O₃: {olcum_b2o3}%")
print(f"Na₂O: {olcum_na2o}%")
print(f"SO₄: {olcum_so4} ppm")
print(f"Fe: {olcum_fe} ppm")
```

```python
urun_spektleri
```

```python
spekt_b2o3 = urun_spektleri['Etibor-48']['B2O3']
b2o3_uygun = (olcum_b2o3 >= spekt_b2o3['min']) and (olcum_b2o3 <= spekt_b2o3['max'])

print(f"\nB₂O₃ Kontrolü:")
print(f"  Spekt: {spekt_b2o3['min']}-{spekt_b2o3['max']}{spekt_b2o3['birim']}")
print(f"  Ölçüm: {olcum_b2o3}{spekt_b2o3['birim']}")
print(f"  Durum: {'✓ UYGUN' if b2o3_uygun else '✗ UYGUN DEĞİL'}")
```

```python
# Diğer Parametreler
spekt_na2o = urun_spektleri[urun_adi]["Na2O"]
na2o_uygun = (olcum_na2o >= spekt_na2o["min"]) and (olcum_na2o <= spekt_na2o["max"])

spekt_so4 = urun_spektleri[urun_adi]["SO4"]
so4_uygun = olcum_so4 <= spekt_so4["max"]

spekt_fe = urun_spektleri[urun_adi]["Fe"]
fe_uygun = olcum_fe <= spekt_fe["max"]

print(f"Na₂O: {'✓' if na2o_uygun else '✗'}")
print(f"SO₄: {'✓' if so4_uygun else '✗'}")
print(f"Fe: {'✓' if fe_uygun else '✗'}")

tum_uygun = b2o3_uygun and na2o_uygun and so4_uygun and fe_uygun
print(f"\n{'='*40}")
print(f"SONUÇ: {'✓ TÜM TESTLER UYGUN' if tum_uygun else '✗ UYGUNSUZLUK'}")
print(f"{'='*40}")
```
